# Income Tax Act Sections Downloader using Pyautogui

```python
import pyautogui as pag, time

# Sections Download 
pag.click(140, 19, duration = 0.5) # click on 1st tab = 140, 19
time.sleep(2)
pag.click(555, 561, duration = 0.5) # click on 1st section = 554, 529
for i in range(1, 930):
    time.sleep(5)
    pag.click(1762, 206) # click to print current section = 1762, 206
    time.sleep(2)
    pag.click(810, 843) # click to print button to save the pdf = 799, 843
    time.sleep(0.5)
    # pag.click(291, 637, duration = 1) # click and write the name of the file = 291, 637
    pag.press("backspace") # click backspace to clear the current name of the file
    pag.write(f"section-{i}") # click and write the name of the file = 291, 637
    time.sleep(0.3)
    pag.click(849, 766) # click to save the pdf = 849, 766
    pag.click(989, 18) # click to close the new tab
    time.sleep(0.3)
    pag.click(1705, 214, duration = 1) # click to next section = 1705, 214
        # 8th click = repeat
```
